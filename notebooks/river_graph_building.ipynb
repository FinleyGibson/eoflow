```python
from shapely.geometry.polygon import Polygon
import matplotlib.pyplot as plt
from eoflow.folium import add_polygon_to_folium
import geopandas as gpd
from shapely.geometry import LineString, MultiLineString
from eoflow.rivers import get_river_network_from_shape, calculate_graph_length_meters
import random
import networkx as nx

import folium
```

```python
# define polygon of interest
POLY = Polygon((
    (-3.5339, 50.7284), 
    (-3.5252, 50.7234), 
    (-3.5252, 50.7134), 
    (-3.5339, 50.7084), 
    (-3.5425, 50.7134), 
    (-3.5425, 50.7234), 
    (-3.5339, 50.7284)
    ))
```

```python
map_kwargs = {
    "scroll_wheel_zoom":False, 
    "height":"500px", 
    "width":"800px"
}
m = add_polygon_to_folium(POLY, map_kwargs=map_kwargs)
m
```

```python
rivers, graph = get_river_network_from_shape(POLY, return_graph=True)
```

```python
def random_color():
    return f"#{random.randint(0, 0xFFFFFF):06x}"

folium.GeoJson(
    rivers,
    name="Rivers",
    style_function=lambda feature: {
        "color": random_color(),
        "weight": 2
    },
).add_to(m)
m
```

```python
plt.figure(figsize=(6, 6))
nx.draw(graph, with_labels=True, node_color='skyblue', edge_color='gray', node_size=50, font_size=7)
plt.show()
```
