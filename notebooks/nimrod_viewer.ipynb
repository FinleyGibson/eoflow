```python
from pathlib import Path
import pandas as pd
import geopandas as gpd

# Optional: geopandas for spatial files
try:
except Exception:
    gpd = None

data_dir = Path("/home/finley/Work/RDS/projects/enforce/repos/eoflow/data/nimrod/2026")

tabular_exts = {".csv", ".parquet", ".feather", ".json", ".xlsx"}
geo_exts = {".gpkg", ".shp", ".geojson", ".json"}

dataframes = []
geodataframes = []

for fp in data_dir.rglob("*"):
    if not fp.is_file():
        continue

    ext = fp.suffix.lower()

    try:
        # Tabular files -> pandas
        if ext in tabular_exts:
            if ext == ".csv":
                dataframes.append(pd.read_csv(fp))
            elif ext == ".parquet":
                dataframes.append(pd.read_parquet(fp))
            elif ext == ".feather":
                dataframes.append(pd.read_feather(fp))
            elif ext == ".json":
                dataframes.append(pd.read_json(fp))
            elif ext == ".xlsx":
                dataframes.append(pd.read_excel(fp))

        # Geospatial files -> geopandas
        elif gpd is not None and ext in geo_exts:
            geodataframes.append(gpd.read_file(fp))

    except Exception as e:
        print(f"Skipped {fp}: {e}")

# Final outputs (use these in later cells)
df = pd.concat(dataframes, ignore_index=True) if dataframes else pd.DataFrame()
gdf = pd.concat(geodataframes, ignore_index=True) if geodataframes else (
    gpd.GeoDataFrame() if gpd is not None else None
)

print(f"Loaded pandas rows: {len(df):,}")
if gdf is not None:
    print(f"Loaded geopandas rows: {len(gdf):,}")
else:
    print("geopandas not available.")
```
